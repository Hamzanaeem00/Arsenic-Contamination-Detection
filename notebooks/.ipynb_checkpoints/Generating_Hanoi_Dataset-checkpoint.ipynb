# Using Hanoi Network

CREATING 100 SCENARIOS AND DATASET

```python
import random
import numpy as np
import pandas as pd

from epyt_flow.simulation import (
    ScenarioConfig,
    ScenarioSimulator,
    EpanetConstants
)

from epyt_flow.simulation.events import SpeciesInjectionEvent
from epyt_flow.utils import to_seconds


#########################################################
# USER SETTINGS
#########################################################

NUM_SCENARIOS = 100

SIMULATION_DAYS = 21

CONTAMINATION_DURATION = 1

MIN_CONCENTRATION = 20

MAX_CONCENTRATION = 200

OUTPUT_FILE = "../data/dataset_hanoi.csv"
chlorine_sensor_locations = [

    "3",      # Main junction

    "8",      # Upper branch

    "13",     # End of upper branch

    "17",     # Middle branch

    "22",     # End of branch 1

    "24",     # Branch 2

    "27",     # End of branch 2

    "29",     # Branch 3

    "32"      # End of branch 3

]


#########################################################
# LOAD HANOI NETWORK
#########################################################

config = ScenarioConfig(

    f_inp_in="../HanoiNetwork/Hanoi.inp",

    f_msx_in="../starter-code/arsenic_contamination.msx"

)

all_data = []

print("="*60)
print("Generating Hanoi Dataset")
print("="*60)


#########################################################
# MAIN LOOP
#########################################################

for scenario in range(NUM_SCENARIOS):

    print(f"\nScenario {scenario+1}/{NUM_SCENARIOS}")

    with ScenarioSimulator(
            scenario_config=config) as sim:

        sim.set_general_parameters(

            simulation_duration=to_seconds(days=SIMULATION_DAYS)

        )

        #########################################################
        # Get all junctions except the reservoir (Node 1)
        #########################################################

        all_nodes = [
        node
        for node in sim.sensor_config.nodes
        if int(node) != 1
        ]

        print("Total Nodes:", len(all_nodes))

        sim.set_bulk_species_node_sensors({

            "Chlorine": chlorine_sensor_locations,

            "AsIII": all_nodes

        })

        #########################################################
        # RANDOM CONTAMINATION EVENT
        #########################################################

        injection_node = random.choice(all_nodes)

        concentration = random.uniform(
            MIN_CONCENTRATION,
            MAX_CONCENTRATION
        )

        start_day = random.randint(
            1,
            SIMULATION_DAYS - CONTAMINATION_DURATION - 1
        )

        end_day = start_day + CONTAMINATION_DURATION

        print(f"Injection Node      : {injection_node}")
        print(f"Injection Amount    : {concentration:.2f} mg/L")
        print(f"Start Day           : {start_day}")
        print(f"End Day             : {end_day}")

        #########################################################
        # CREATE CONTAMINATION EVENT
        #########################################################

        contamination_event = SpeciesInjectionEvent(

            species_id="AsIII",

            node_id=str(injection_node),

            profile=np.array([concentration]),

            source_type=EpanetConstants.EN_MASS,

            start_time=to_seconds(days=start_day),

            end_time=to_seconds(days=end_day)

        )

        #########################################################
        # ADD EVENT TO SIMULATION
        #########################################################

        sim.add_system_event(contamination_event)

        #########################################################
        # RUN SIMULATION
        #########################################################

        print("Running simulation...")

        scada_data = sim.run_simulation()

        #########################################################
        # CONVERT TO DATAFRAME
        #########################################################

        df = scada_data.to_pandas_dataframe()

        print("Rows :", len(df))
        print("Columns :", len(df.columns))

        ######################################################
        # ADD METADATA
        ######################################################

        df["Scenario"] = scenario + 1

        df["InjectionNode"] = injection_node

        df["InjectionAmount"] = concentration

        df["StartDay"] = start_day

        df["EndDay"] = end_day

        ######################################################
        # CREATE ATTACK LABEL
        ######################################################

        # Simulation information
        total_steps = len(df)

        total_simulation_seconds = SIMULATION_DAYS * 24 * 60 * 60

        seconds_per_step = total_simulation_seconds / (total_steps - 1)

        attack_start = to_seconds(days=start_day)

        attack_end = to_seconds(days=end_day)

        df["Attack"] = 0 # Initialize Attack column

        for i in range(total_steps):

            current_time = i * seconds_per_step

            if attack_start <= current_time <= attack_end:

                df.loc[i, "Attack"] = 1

        ######################################################
        # OPTIONAL:
        # Remove Arsenic Columns
        ######################################################

        arsenic_cols = [c for c in df.columns if "AsIII" in c]

        # Update attack label based on actual arsenic presence
        df["Attack"] = (df[arsenic_cols].max(axis=1) > 0).astype(int)

        df.drop(columns=arsenic_cols, inplace=True)

        # We do NOT want the ML model
        # to directly see the arsenic values.


        ######################################################
        # APPEND TO DATASET
        ######################################################

        all_data.append(df)

        print("Scenario Complete.")

##############################################################
# COMBINE ALL SCENARIOS
##############################################################

print("\n")
print("=" * 60)
print("Combining all scenarios...")
print("=" * 60)

dataset = pd.concat(all_data, ignore_index=True)

##############################################################
# REORDER COLUMNS
##############################################################

metadata_columns = [
    "Scenario",
    "InjectionNode",
    "InjectionAmount",
    "StartDay",
    "EndDay",
    "Attack"
]

other_columns = [
    col for col in dataset.columns
    if col not in metadata_columns
]

dataset = dataset[
    metadata_columns + other_columns
]

##############################################################
# SAVE DATASET
##############################################################

dataset.to_csv(
    OUTPUT_FILE,
    index=False
)

##############################################################
# DATASET SUMMARY
##############################################################

print("\n")
print("=" * 60)
print("DATASET CREATED SUCCESSFULLY")
print("=" * 60)

print(f"Number of scenarios : {NUM_SCENARIOS}")
print(f"Total rows          : {len(dataset)}")
print(f"Total columns       : {len(dataset.columns)}")

print("\nAttack Distribution")

print(dataset["Attack"].value_counts())

print("\nFirst 5 Rows")

print(dataset.head())

print("\nDataset saved as")

print(OUTPUT_FILE)

print("=" * 60)


df = scada_data.to_pandas_dataframe()

#########################################################
# ADD SCENARIO INFORMATION
#########################################################

df["Scenario"] = scenario + 1

df["InjectionNode"] = injection_node

df["InjectionAmount"] = concentration

df["StartDay"] = start_day

df["EndDay"] = end_day


#########################################################
# CREATE ATTACK LABEL
#########################################################

# Find all arsenic columns
arsenic_columns = [

    col

    for col in df.columns

    if "AsIII" in col

]

# Attack = 1 if arsenic is detected anywhere
df["Attack"] = (

    df[arsenic_columns]

    .max(axis=1)

    > 0

).astype(int)


#########################################################
# REMOVE ARSENIC COLUMNS
#########################################################

df.drop(

    columns=arsenic_columns,

    inplace=True

)


#########################################################
# APPEND TO DATASET
#########################################################

all_data.append(df)

print(f"Rows Generated : {len(df)}")

print("Scenario Complete!")

#########################################################
# COMBINE ALL SCENARIOS
#########################################################

dataset = pd.concat(

    all_data,

    ignore_index=True

)

#########################################################
# REORDER COLUMNS
#########################################################

metadata = [

    "Scenario",

    "InjectionNode",

    "InjectionAmount",

    "StartDay",

    "EndDay",

    "Attack"

]

other_columns = [

    c

    for c in dataset.columns

    if c not in metadata

]

dataset = dataset[

    metadata + other_columns

]


#########################################################
# SAVE DATASET
#########################################################


dataset.to_csv(

    OUTPUT_FILE,

    index=False

)


#########################################################
# PRINT SUMMARY
#########################################################

print("\n")

print("="*60)

print("DATASET CREATED SUCCESSFULLY")

print("="*60)

print(f"Total Scenarios : {NUM_SCENARIOS}")

print(f"Total Rows      : {len(dataset)}")

print(f"Total Columns   : {len(dataset.columns)}")

print("\nAttack Distribution")

print(dataset["Attack"].value_counts())

print("\nDataset saved as")

print(OUTPUT_FILE)

print("="*60)


```
